#**Ejercicio (Bootstrap):**

Esta es una muestra aleatoria de 12 tiempos de entrega:
$$35,42,38,40,45,37,39,41,44,36,43,40$$



Instrucciones:


1.   Use la muestra originalde 12 observaciones y genere 1000 muestras Bootstrap ($F_e(x)$)
2.   Para cada muestra B calcule la media
3.   Al final:

*   Obtenga la media de las 1000 muestras
*   Calcule la desviación estándar
*   Determine los porcentiles $25\% $ y $97.5\%$ para construir el intervalo de confianza B del $95\%$








```python
#Importamos las librerias
import numpy as np
import random as r
import matplotlib.pyplot as plt
```

```python
datos =np.array([35,42,38,40,45,37,39,41,44,36,43,40])  #datos del problema
n=len(datos)    #tamaño de la muestra
B=100   #No.de remuestreos bootstrap
media_b = []  #lista para almacenar las medias de las muestras
```

```python
for i in range(B):
    muestra_b = []  #lista para almacenar las muestras bootstrap

    # generar muestra bootstrap
    for j in range(n):
        x = r.choice(datos)   #elegimos un dato al azar de 'datos' con reemplazo
        muestra_b.append(x)  #lo agregamos a la lista
    media = np.mean(muestra_b)  #calculamos la media de la muestra bootstrap
    media_b.append(media)  #agregamos la media a la lista de medias

media_b = np.array(media_b)   #Convertimos la lista en un arreglo para usar las funciones de numpy

#Resultados
media_final = np.mean(media_b)  #media
desviacion = np.std(media_b)    #desviación
p1 = np.percentile(media_b,2.5) #percentil
p2 = np.percentile(media_b,97.5)

#Imprimimos los resultados
print("Media Bootstrap =",media_final)
print("Desviacion =",desviacion)
print("Intervalo 95%")
print("[",p1,",",p2,"]")
```

```python
plt.hist(media_b, bins=20)
plt.xlabel("Media muestral")
plt.ylabel("Frecuencia")
plt.title("Distribucion Bootstrap de las medias")
plt.show()
```
