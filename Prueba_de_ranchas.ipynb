#**Prueba de independencia o rachas**

La *prueba de rachas* es un método que se utiliza para determinar si una secuencia datos ocurre de manera aleatoria, esta prueba analiza el número de rachas representadas por + y -.

La hipotesis nula es:
$$ H_0: \text{Los números de la muestra} \ {x_i} \ \text{son independientes}$$
$$ H_1: \text{Los números de la muestra} \ {x_i} \ \text{no son independientes}$$

```python
import numpy as np
import pandas as pd
import scipy.stats as stats
```

```python
datos= [0.78961, 0.76086, 0.80548, 0.58518, 0.89898, 0.28269, 0.38618, 0.79982, 0.58962, 0.69623, 0.29931, 0.57410,0.24000,0.93655, 0.54325,0.58244,0.23949,0.19962,0.19147,0.22287,0.05230, 0.12079, 0.82654, 0.98611, 0.57880, 0.73059, 0.76910, 0.45679, 0.33216, 0.17028, 0.30861, 0.26593, 0.65559, 0.88809, 0.62400, 0.85853, 0.52559, 0.65002, 0.40644, 0.07281, 0.10699, 0.27738, 0.29453, 0.34488, 0.67621, 0.70119, 0.68334, 0.21631, 0.03185, 0.05475, 0.83358, 0.85903, 0.38507, 0.81772, 0.09133, 0.88752, 0.33381, 0.74579, 0.08128, 0.64183, 0.55877, 0.65726, 0.20852, 0.34358, 0.05010, 0.18284, 0.55170, 0.87616, 0.61168, 0.91512, 0.51781, 0.43308, 0.90829, 0.36982, 0.41678, 0.33729, 0.49383, 0.79113, 0.73435, 0.44267, 0.14151, 0.79269, 0.42989,0.11537, 0.00121, 0.49962, 0.10850, 0.55743, 0.09264, 0.76262, 0.03272, 0.35286, 0.94187, 0.19904,0.33954, 0.15506, 0.75103, 0.63453, 0.22724, 0.72102]
```

 ## **Formulas**
 $$ \mu_{C_0}= \frac{2n-1}{3}$$
 $$σ ^2 _{C_0}= \frac{16n-29}{90}$$
 $$ z_0= \frac{C_0-\mu_{C_0}}{\sigma_{C_0}}$$

```python

n=len(datos)

def prueba_de_rachas(datos): #Definimos la funcion pruba de rachas que necesita como parametro los datos
  media= (2*n-1) / 3
  varianza_cuadrada=(16*n-29) / 90     #Formulas de la media y varianza cuadrada
  varianza=np.sqrt(varianza_cuadrada)

  rachas=1 #Iniciamos el conteo
  signos=[] #Arreglo para guardar los signos de las rachas

  for i in range(1,n): #Creamos un ciclo donde separemos los signos
    if datos[i-1]<datos[i]:
      signos.append('+')
    else:
      signos.append('-')

  for i in range(1,n-1): #Creamos un ciclo donde hacemos el conteo de las rachas
    if signos[i-1]!=signos[i]: #Si los signos son diferentes sumamos uno a la racha
      rachas+=1

  return rachas,media,varianza,varianza_cuadrada

rachas, media, varianza, varianza_cuadrada = prueba_de_rachas(datos)

alpha = 0.05 # Definimos alpha
z_0=(rachas - media) / varianza #Formula de z_0
z_critico=stats.norm.ppf(1-(alpha/2)) #Calculo del z critico

# Imprimimos los resoltados
print('------------ Resultados Prueba de Rachas -----------')
print('Media (M_c0):', round(media,3))
print('Varianza cuadrada (Sigma_c0):',round(varianza_cuadrada,3))
print('Z_0:',round(z_0,3))
print('Z_critico:', round(z_critico,3))

if abs(z_0) > z_critico:
  print('Rechazamos H_0: Los números NO son independientes')
else:
  print('Aceptamos H_0: Los números son independientes (aleatorios)')
```
