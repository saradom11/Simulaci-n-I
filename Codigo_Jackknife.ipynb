#**Método Jackknife**

El método de Jackknife es una técnica de remuestreo utilizada principalmente para estimar el sesgo y el error estándar de un estadístico (como una media o varianza) cuando la distribución teórica de los datos es desconocida.

Esta es una muestra aleatoria de 12 tiempos de entrega:
$$35,42,38,40,45,37,39,41,44,36,43,40$$



```python
import numpy as np

datos = np.array([35,42,38,40,45,37,39,41,44,36,43,40]) #datos del problema
n = len(datos)   #tamaño de muestra
medias = []   #quí guardamos las medias

# Quitar un dato cada vez
for i in range(n):
    muestra = np.delete(datos,i)   #quitamos el dato i
    media = np.mean(muestra)    #media dejando uno fuera
    medias.append(media)

medias = np.array(medias)
media_jack = np.mean(medias)    #promedio de todas las medias


error = np.sqrt((n-1)/n * np.sum((medias-media_jack)**2))   #error jackknife

#Imprimimos los resultados
print("Medias Jackknife:",medias)
print("Media final:",media_jack)
print(f"Error estándar: {error:.4f}")
```
